```python
df = df.drop(['Posted On', 'Floor', 'Area Locality'], axis=1)
df
```

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.tree import DecisionTreeRegressor
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split

import sklearn.metrics as metrics
import category_encoders as ce
from sklearn.preprocessing import MinMaxScaler
```

```python
categorical_cols = df.select_dtypes(include = 'object').columns
categorical_cols
```

```python
for col in categorical_cols:
    print(f"{col} : {df[col].nunique()}")
```

```python
x=df.drop(columns='Rent')
y=df['Rent']
```

```python
encoder = ce.LeaveOneOutEncoder(return_df=True)
x=encoder.fit_transform(x,y)
```

```python
from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler()
x = scaler.fit_transform(x)
```

```python
x_df = pd.DataFrame(x)
x_df
```

```python
x_train, x_test, y_train, y_test = train_test_split(x,y, test_size=0.20, random_state=40)
```

```python
x_train.shape, x_test.shape
```

```python
model_rfr = RandomForestRegressor(n_estimators=7)
model_dt = DecisionTreeRegressor()
model_lr = LinearRegression()

models = [model_rfr, model_dt, model_lr]
```

```python
for model in models :
    print(f"fitting model: {model}")
    model.fit(x_train,y_train)
```

```python
for model in models:
    print(f"score of {model}: ")
    print(f"  for training data: {model.score(x_train, y_train)}")
    print(f"  for test data: {model.score(x_test, y_test)}")
```

```python
def regression_results(y_true, y_pred):
    explained_variance = metrics.explained_variance_score(y_true, y_pred)
    mean_absolute_error = metrics.mean_absolute_error(y_true, y_pred)
    mse = metrics.mean_squared_error(y_true, y_pred)
    median_absolute_error = metrics.median_absolute_error(y_true, y_pred)
    r2 = metrics.r2_score(y_true, y_pred)
    print('explained_variance: ', round(explained_variance,4))    
    print('r2: ', round(r2,4))
    print('MAE: ', round(mean_absolute_error,4))
    print('MSE: ', round(mse,4))
    print('RMSE: ', round(np.sqrt(mse),4))
    print('Median absolute error: ',round(median_absolute_error,4))
```

```python
for model in models[:]:
    y_predicted = model.predict(x_test)

    print(f"Report: {model}")
    print(f"{regression_results(y_test, y_predicted)}\n")
```
